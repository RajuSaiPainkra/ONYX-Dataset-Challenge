```python
data.dtypes # Checking Data Types
```

```python
# Classification the Column in Numerical, Ordinal, Categorical Datasets
```

```python
def categorize_columns(data):
    numerical_colm = []
    categorical_colm = []
    ordinal_colm = []

    for column in data.columns:
        if data[column].dtype in ['int64', 'float64']: # for Numerical Columns
            numerical_colm.append(column)
        elif data[column].dtype == "object": # for Categorical Columns
            categorical_colm.append(column)
        else:
            ordinal_colm.append(column)
    return numerical_colm, categorical_colm, ordinal_colm

# categorize the column
numerical_colm, categorical_colm, ordinal_colm = categorize_columns(data)

# print the result
print("Numerical Columns : ", numerical_colm)
print("\nCategorical Columns : ", categorical_colm)
print("\nOrdinal Columns : ", ordinal_colm )
```

## Data Distribution

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Visualizing distribution of all numerical columns
data.hist(figsize=(10, 8), bins=20, edgecolor='k')
plt.suptitle('Histograms of Numerical Features')
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Visualizing distribution of all numerical columns
num_columns = data.select_dtypes(include=['int64', 'float64']).columns  # Select numerical columns
num_features = len(num_columns)  # Count of numerical features

# Create subplots with an appropriate grid size
rows = (num_features // 3) + (1 if num_features % 3 != 0 else 0)  # Calculate rows for a grid of 3 columns
fig, axes = plt.subplots(rows, 3, figsize=(15, 5 * rows))  # Adjust figure size for better spacing
axes = axes.flatten()  # Flatten to iterate over each subplot easily

# Plot each numerical feature
for i, col in enumerate(num_columns):
    sns.histplot(data[col], bins=20, kde=False, ax=axes[i], color='skyblue', edgecolor='k')
    axes[i].set_title(f'Distribution of {col}')
    axes[i].set_xlabel(col)
    axes[i].set_ylabel('Frequency')

# Remove empty subplots (if any)
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])  # Delete unused axes

# Add spacing and adjust layout
plt.tight_layout()
plt.suptitle('Histograms of Numerical Features', y=1.02, fontsize=16)  # Add a main title
plt.show()
```
